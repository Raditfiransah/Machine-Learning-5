# Ekstraksi Fitur Teks Item

Uraian item pada struk diubah menjadi vektor TF-IDF, lalu direduksi dengan SVD.

## 1. Muat Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

CSV = '../data/Transactions.csv'
df = pd.read_csv(CSV)
df.head()
```

## 2. TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(token_pattern=r'[^,]+', lowercase=True)
X_tfidf = tfidf.fit_transform(df['Items'])
print('Matriks TF-IDF:', X_tfidf.shape)
```

## 3. Reduksi Dimensi dengan SVD

```python
from sklearn.decomposition import TruncatedSVD

svd = TruncatedSVD(n_components=5, random_state=42)
X_svd = svd.fit_transform(X_tfidf)
print('Explained variance:', svd.explained_variance_ratio_.round(3))
print('Total:', svd.explained_variance_ratio_.sum().round(3))
```

## 4. Contoh Hasil

```python
hasil = pd.DataFrame(X_svd, columns=[f'komponen_{i+1}' for i in range(5)])
hasil['Items'] = df['Items']
hasil.head()
```
