## Langkah 0 - Persiapan

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

## Langkah 1 - Load Data

```python
dpath = '../data/Titanic-Dataset-fixed.csv'
df = pd.read_csv(dpath)
df.head()
```

## Langkah 2 - Slice Data

Fitur yang digunakan: `Survived`, `Pclass`, `Age`, `Sex`, `Cabin`.

```python
df = df[['Survived', 'Pclass', 'Age', 'Sex', 'Cabin']].copy()
df.head()
```

## Langkah 3 - Encoding

Label Encoding pada kolom kategorikal `Sex` dan `Cabin`.

```python
le = LabelEncoder()  # membuat objek dari LabelEncoder
df['Sex'] = le.fit_transform(df['Sex'])  # proses encoding
df['Cabin'] = le.fit_transform(df['Cabin'])  # proses encoding
```

## Langkah 4 - Verifikasi Hasil

```python
df.head()
```

## Langkah 5 - Standarisasi

Standardisasi pada kolom `Age`.

```python
std = StandardScaler()
df['Age'] = std.fit_transform(df[['Age']])
```
