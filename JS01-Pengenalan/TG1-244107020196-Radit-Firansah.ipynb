### 2. Kegunaan masing-masing library

1. **PyPREP**
   Digunakan untuk **preprocessing data EEG**, seperti membersihkan noise dan mendeteksi channel EEG yang bermasalah.

2. **SciPy**
   Library Python untuk **komputasi ilmiah**, termasuk pengolahan sinyal, statistik, optimasi, dan operasi matematika.

3. **Weights & Biases (wandb)**
   Digunakan untuk **melacak dan memonitor eksperimen machine learning**, seperti loss, akurasi, parameter, dan hasil pelatihan model.

4. **pyECG**
   Digunakan untuk **pengolahan dan analisis sinyal ECG (electrocardiogram)**, misalnya untuk membantu menganalisis sinyal aktivitas listrik jantung.

### 1. **Tindakan yang melanggar etika dan hukum dalam penggunaan AI:**
   Contohnya adalah menggunakan AI untuk membuat dan menyebarkan **deepfake** seseorang tanpa izin, terutama untuk penipuan atau pencemaran nama baik. Hal ini melanggar etika karena merugikan dan menipu orang lain, serta dapat melanggar hukum terkait data pribadi, penipuan, atau pencemaran nama baik.
   **Referensi:** UU No. 27 Tahun 2022 tentang Perlindungan Data Pribadi; UNESCO, *Recommendation on the Ethics of Artificial Intelligence*.

### 2. **Dampak energi dan lingkungan dari AI:**
   AI membutuhkan pusat data yang menggunakan listrik dalam jumlah besar dan juga membutuhkan air untuk sistem pendinginan. Peningkatan penggunaan AI dapat meningkatkan konsumsi energi dan emisi karbon. Menurut saya, cara mengatasinya adalah menggunakan **energi terbarukan**, meningkatkan efisiensi model AI, serta menggunakan perangkat keras dan pusat data yang hemat energi.
   **Referensi:** International Energy Agency (IEA), *Energy and AI (2025)*; UNESCO, *Recommendation on the Ethics of Artificial Intelligence*.
